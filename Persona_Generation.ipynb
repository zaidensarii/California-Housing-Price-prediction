```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import FactorAnalysis
from sklearn.metrics.pairwise import euclidean_distances

def preprocess_text(text):

    return text.lower()

def calculate_similarity_with_factor_analysis(ideal_answer, actual_answer):

    ideal_answer = preprocess_text(ideal_answer)
    actual_answer = preprocess_text(actual_answer)


    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([ideal_answer, actual_answer])


    factor_analysis = FactorAnalysis(n_components=2)
    factors = factor_analysis.fit_transform(tfidf_matrix.toarray())


    distance = euclidean_distances(factors[0:1], factors[1:2])


    similarity_score = 1 / (1 + distance[0][0])

    return similarity_score

ideal_answer = "Memory management in iOS is handled using ARC, which automatically tracks and manages the memory of objects. When an object's reference count drops to zero, the memory is freed. To optimize memory usage, I avoid retain cycles by using weak or unowned references for closures and delegate patterns. Profiling tools like Instruments help identify memory leaks and optimize usage."
actual_answer = "I understand that ARC helps manage memory by automatically freeing up objects when they're no longer needed. I usually follow best practices like using weak references for delegates, but I haven't had much experience with more advanced memory management techniques or troubleshooting memory issues."

similarity_score = calculate_similarity_with_factor_analysis(ideal_answer, actual_answer)
print(f"Similarity Score using Factor Analysis: {similarity_score:.2f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def compare_answers(ideal_answer, actual_answer):

    vectorizer = TfidfVectorizer()


    tfidf_matrix = vectorizer.fit_transform([ideal_answer, actual_answer])


    similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

    return similarity[0][0]


ideal_answer = "Memory management in iOS is handled using ARC, which automatically tracks and manages the memory of objects. When an object's reference count drops to zero, the memory is freed. To optimize memory usage, I avoid retain cycles by using weak or unowned references for closures and delegate patterns. Profiling tools like Instruments help identify memory leaks and optimize usage."
actual_answer = "I understand that ARC helps manage memory by automatically freeing up objects when they're no longer needed. I usually follow best practices like using weak references for delegates, but I haven't had much experience with more advanced memory management techniques or troubleshooting memory issues."

similarity_score = compare_answers(ideal_answer, actual_answer)
print(f"Similarity Score: {similarity_score:.2f}")
```
